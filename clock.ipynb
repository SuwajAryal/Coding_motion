```python
import pygame
import sys

# --- Setup ---
pygame.init()

WIDTH, HEIGHT = 640, 480
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Pygame Setup")
clock = pygame.time.Clock()

WHITE = (255, 255, 255)
BLUE  = (30, 100, 200)

# --- Game loop ---
running = True
while running:
    # 1. Handle events
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    # 2. Update game state (nothing to update yet — static scene)

    # 3. Draw
    screen.fill(WHITE)
    pygame.draw.circle(screen, BLUE, (WIDTH // 2, HEIGHT // 2), 60)

    # 4. Flip display
    pygame.display.flip()

    # 5. Cap frame rate
    clock.tick(60)

pygame.quit()
sys.exit()
```
