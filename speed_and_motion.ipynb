```python
#applying the gravity concept
# very basic 
# change in shape but still missing ....
# lacking funtions 
# learn to draw the shapes , ... redrop, not too interactive...


import pygame
import sys

# --- Setup ---
pygame.init()

WIDTH, HEIGHT = 640, 480
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Dropping Object - Choose Shape")
clock = pygame.time.Clock()

WHITE = (255, 255, 255)
BLACK = (20, 20, 20)

SIZE = 30
START_Y = 50          # remember the starting height so we can re-drop

x = WIDTH // 2
y = START_Y
vy = 0
gravity = 0.5

shape = "circle"      # press 1 / 2 / 3 to change shape
                       # press SPACE to drop again

# --- Game loop ---
running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
        elif event.type == pygame.KEYDOWN:
            if event.key == pygame.K_1:
                shape = "circle"
            elif event.key == pygame.K_2:
                shape = "square"
            elif event.key == pygame.K_3:
                shape = "triangle"
            elif event.key == pygame.K_SPACE:
                y = START_Y     # reset position back to top
                vy = 0          # reset speed back to zero

    # Apply gravity
    vy += gravity
    y += vy

    # Stop at the ground
    if y + SIZE >= HEIGHT:
        y = HEIGHT - SIZE
        vy = 0

    # Draw
    screen.fill(WHITE)

    if shape == "circle":
        pygame.draw.circle(screen, BLACK, (int(x), int(y)), SIZE)
    elif shape == "square":
        pygame.draw.rect(screen, BLACK, (int(x - SIZE), int(y - SIZE), SIZE * 2, SIZE * 2))
    elif shape == "triangle":
        points = [(x, y - SIZE), (x - SIZE, y + SIZE), (x + SIZE, y + SIZE)]
        pygame.draw.polygon(screen, BLACK, points)

    pygame.display.flip()
    clock.tick(60)

pygame.quit()
sys.exit()
```
