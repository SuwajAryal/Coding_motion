```python
import pygame
import sys

# --- Setup ---
pygame.init()

WIDTH, HEIGHT = 640, 480
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Rolling Tire")
clock = pygame.time.Clock()

WHITE = (255, 255, 255)
BLACK = (20, 20, 20)
GRAY  = (120, 120, 120)

RADIUS = 40
x1, y1 = 50, HEIGHT // 2   # starting position (left side)
speed = 1                # pixels moved per frame — this is the "velocity" 
# 20 is too much , can't even see in screen, but why?? 


#adding another circle 
RADIUS = 40
x2, y2 = 200, HEIGHT // 2  # starting position (left side)
speed1 = 1  
speed2 = 2  #lets try with different speed ...



# --- Game loop ---
running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    # Move the tire to the right
    x1 += speed1
    x2 += speed2

    # Draw
    screen.fill(WHITE)
    pygame.draw.circle(screen, BLACK, (int(x1), y1), RADIUS)        # outer tire
    pygame.draw.circle(screen, GRAY, (int(x1), y1), RADIUS // 3)    # inner hub

    #another tire
    pygame.draw.circle(screen, BLACK, (int(x2), y2), RADIUS)
    pygame.draw.circle(screen, GRAY, (int(x2), y2), RADIUS // 3)

    pygame.display.flip()
    clock.tick(60)

pygame.quit()
sys.exit()
```
